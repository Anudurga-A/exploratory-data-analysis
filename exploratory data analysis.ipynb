```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    'Student': ['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H', 'I', 'J'],
    'Study_Hours': [2, 4, 5, 3, 6, 8, 7, 4, 9, 5],
    'Attendance': [60, 70, 75, 65, 80, 95, 90, 70, 98, 85],
    'Marks': [45, 55, 60, 50, 70, 90, 85, 58, 95, 68]
}

df = pd.DataFrame(data)

df
```

```python
df.head()
```

```python
print("Dataset Shape:", df.shape)

print("\nDataset Information:")
df.info()
```

```python
df.describe()
```

```python
print("Missing Values:")
print(df.isnull().sum())
```

```python
print("Duplicate Rows:", df.duplicated().sum())

df = df.drop_duplicates()

print("Dataset Shape After Removing Duplicates:", df.shape)
```

```python
plt.figure(figsize=(7, 5))

plt.hist(df['Marks'], bins=5, edgecolor='black')

plt.title('Distribution of Student Marks')
plt.xlabel('Marks')
plt.ylabel('Number of Students')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.bar(df['Student'], df['Marks'])

plt.title('Student-wise Marks')
plt.xlabel('Student')
plt.ylabel('Marks')

plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(df['Study_Hours'], df['Marks'])

plt.title('Study Hours vs Marks')
plt.xlabel('Study Hours')
plt.ylabel('Marks')

plt.show()
```

```python
plt.figure(figsize=(7, 5))

plt.boxplot(df['Marks'])

plt.title('Box Plot of Student Marks')
plt.ylabel('Marks')

plt.show()
```
